# Tool · Keep only strokes and extend them by one frame

**What this notebook does:** It works on the tables made by the *ELAN tier exporter* tool. From each table it
keeps only the annotations with one label (by default `S`, for stroke) and makes each of them one video frame
(33 ms) longer, so the last frame of the stroke is included.

| You need | You get |
|---|---|
| Tables from `elan_tier_exporter.ipynb` (`..._annotations.csv`) | For each table, a new one ending in `_extended.csv`. |

*Notebook 3b does this by itself (see its `EXTEND_END_MS` setting), so you don't need this tool for the main workflow.*

## Step 1 · Load the tools

```python
from pathlib import Path
import pandas as pd
```

## Step 2 · Settings

```python
FOLDER = "../Annotations"   # the folder with the exported tables
KEEP_LABEL = "S"            # keep only annotations with this text
EXTEND_MS = 33              # add this many milliseconds to the end of each annotation
```

## Step 3 · Filter and extend

```python
for path in sorted(Path(FOLDER).glob("*_annotations.csv")):
    table = pd.read_csv(path)
    if "Annotation" not in table.columns:
        print(f"Skipping {path.name}: it wasn't made by the ELAN tier exporter")
        continue

    kept = table[table["Annotation"] == KEEP_LABEL].copy()
    kept["End Time"] += EXTEND_MS

    output_path = path.with_name(path.stem + "_extended.csv")
    kept.to_csv(output_path, index=False)
    print(f"{path.name}: kept {len(kept)} of {len(table)} annotations, saved to {output_path.name}")
```
