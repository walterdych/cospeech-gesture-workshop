# Tool · Add participant and task columns to a table

**What this notebook does:** It adds three columns, `subject`, `barename` and `task`, with the same value in every
row, to a results table. This makes it easy to combine tables from many recordings later, for example with EMA
(articulography) data, and still know which row came from which recording.

| You need | You get |
|---|---|
| Any `.csv` table, for example `Annotations/Test_Video_annotations.csv` from Notebook 5 | A copy of the table with the three extra columns, ending in `_completed.csv`. |

## Step 1 · Load the tools

```python
from pathlib import Path
import pandas as pd
```

## Step 2 · Settings

```python
INPUT_FILE = "../Annotations/Test_Video_annotations.csv"

SUBJECT = "P01"          # participant code
BARENAME = "Test_Video"  # name of the recording
TASK = "workshop"        # name of the task
```

## Step 3 · Add the columns and save

```python
table = pd.read_csv(INPUT_FILE)
table["subject"] = SUBJECT
table["barename"] = BARENAME
table["task"] = TASK

output_path = Path(INPUT_FILE).with_name(Path(INPUT_FILE).stem + "_completed.csv")
table.to_csv(output_path, index=False)
print("Saved:", output_path)
table.head()
```
